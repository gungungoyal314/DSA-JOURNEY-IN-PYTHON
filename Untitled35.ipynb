```python
# second largest element in the array

list = [1 , 2, 4, 8 ,9]
largest = list[0]
second = list[0]
for num in list :
  if num > largest:
    second = largest
    largest = num
  elif num > second:
    num!= largest
    second = num
print(second)
```

```python
# maximum element in the list
list = [1 , 2, 4, 8 ,9]
max = list[0]
for num in list:
  if num > max:
    max = num
print(max)
```

```python
# minimum element in the list
list = [1 , 2, 4, 8 ,9]
min = list[0]
for num in list:
  if num < min:
    min = num
print(min)
```

```python
# reverse the array
arr1 = [1,2,3,4,5]
a = 0
b = len(arr1) - 1
while a < b :
  temp = arr1[a]
  arr1 [a] = arr1[b]
  arr1[b] = temp
  a = a+1
  b = b-1
print(arr1)
```

```python
# removing duplicates from the sorted array
arr1 = [1 ,1,2,2,3,4,5,6,7]
result = [ ]
for i in arr1:
  if result == [ ]:
   result.append(i)
  elif result[-1] != i:
   result.append(i)

print(result)
```

```python
# rotate an array by k positions k = 2
k = 2
arr1 = [1,2,3,4,5]
arr1 = arr1[k:] + arr1[:k]
print(arr1)
```
